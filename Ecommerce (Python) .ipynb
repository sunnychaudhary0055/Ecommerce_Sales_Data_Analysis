```python
df["Qty"]=pd.to_numeric(df["Qty"], errors="coerce")         # Qty ko numeric kar dia or NaN ya null values ko errors="coerce" NaN kar dega
```

```python
df["Unit_Price"]=pd.to_numeric(df["Unit_Price"], errors="coerce")
```

```python
df["Discount"]=pd.to_numeric(df["Discount"], errors="coerce")
```

```python
df["Discount"]=df["Discount"].fillna(0)
```

```python
df["Discount"]=df["Discount"].astype("Int64")
```

```python
df["Order_Date"]=pd.to_datetime(df["Order_Date"], format="mixed", dayfirst=True, errors="coerce")
```

```python
df["Delivery_Date"]=pd.to_datetime(df["Delivery_Date"], format = "mixed", dayfirst=True, errors="coerce")
```

```python
df["Delivery_Date"]=df["Delivery_Date"].fillna(df["Order_Date"])
```

```python
df
```

```python
df.info()
```

```python
# Remove extra spaces

df=df.apply(lambda x:x.str.strip().str.title() if x.dtype=="str" else x)
```

```python
df=df[df["Qty"]>=0]
```

```python
df
```

```python
df.describe()
```

```python
# Validate email address

df=df[df["Email"].str.contains('@', na=False)]
```

```python
df
```

```python
df.isnull().sum()
```

# Feature Engineering

```python
# Total Sales

df["Total_Sales"]=df["Qty"]*df["Unit_Price"]
```

```python
df["Total_Sales"]=df["Total_Sales"].round(2)
```

```python
df
```

```python
# Net Amount

df["Net_Amount"]=df["Total_Sales"]-(df["Total_Sales"]*df["Discount"]/100)
```

```python
df["Net_Amount"]=df["Net_Amount"].round(2)
```

```python
df
```

```python
# Profit (Example- 20%)

df["Profit"]=df["Net_Amount"]*0.2
```

```python
df["Profit"]=df["Profit"].round(2)
```

```python
df
```

```python
# Order Month

df["Order_Month"]=df["Order_Date"].dt.month_name()
```

```python
df
```

```python
# Order Year

df["Order_Year"]=df["Order_Date"].dt.year
```

```python
df["Order_Year"].dtype
```

```python
df["Order_Year"]=df["Order_Year"].astype("Int64")
```

```python
df["Order_Year"]=df["Order_Year"].astype("str")
```

```python
df["Order_Year"].dtype
```

```python
df["Order_Year"].isnull().sum()
```

```python
df
```

```python
# WeekDay

df["WeekDay"]=df["Order_Date"].dt.day_name()
```

```python
df
```

```python
col=["Order_Month", "Order_Year", "WeekDay"]
for c in col:
    df[c]=df[c].fillna("Unknown")
```

```python
df["Order_Year"].isnull().sum()
```

```python
df.replace(['N/A', 'NULL', ''], np.nan, inplace=True)
```
